```python
#problem-4
#prompt: Using the original synthetic student data, set StudentID as the index and use .iloc to select Bob, David, and Grace based on their row positions.
import pandas as pd

# Original student data
data = {
    'StudentID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve',
             'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Score': [85, 92, 78, 95, 88, 72, 99, 81, 90, 84],
    'Major': ['Computer Science', 'Physics', 'Mathematics',
              'Computer Science', 'Biology', 'Mathematics',
              'Physics', 'Biology', 'Computer Science', 'Mathematics']
}

# Create the original DataFrame
student_df = pd.DataFrame(data)

print("Original DataFrame:")
print(student_df)


# Reindex the DataFrame using StudentID
df_reindexed = student_df.set_index('StudentID')

print("\nReindexed DataFrame:")
print(df_reindexed)


# Bob is at position 1
# David is at position 3
# Grace is at position 6

# Select Bob, David, and Grace using iloc
selected_students = df_reindexed.iloc[[1, 3, 6]]

print("\nBob, David, and Grace:")
print(selected_students)
```

```python
#problem-5
#prompt: Using the original synthetic student data, set StudentID as the index. Use Boolean indexing with .loc to find Mathematics students, and use .query() to find students with scores greater than 85.
import pandas as pd

# Original student data
data = {
    'StudentID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve',
             'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Score': [85, 92, 78, 95, 88, 72, 99, 81, 90, 84],
    'Major': ['Computer Science', 'Physics', 'Mathematics',
              'Computer Science', 'Biology', 'Mathematics',
              'Physics', 'Biology', 'Computer Science', 'Mathematics']
}

# Create the DataFrame
student_df = pd.DataFrame(data)

print("Original DataFrame:")
print(student_df)


# Set StudentID as the index
df_reindexed = student_df.set_index('StudentID')

print("\nReindexed DataFrame:")
print(df_reindexed)


# Task 1: Create a Boolean Series for Mathematics majors
math_students = df_reindexed['Major'] == 'Mathematics'

print("\nBoolean Series:")
print(math_students)

# Select Mathematics students using .loc
math_result = df_reindexed.loc[math_students]

print("\nMathematics Students:")
print(math_result)


# Task 2: Select students with Score greater than 85 using .query()
high_scores = df_reindexed.query('Score > 85')

print("\nStudents with Scores Greater Than 85:")
print(high_scores)
```

```python
#problem-6
#prompt: Using the original student data, create a function find_top_student_in_major() that filters students by major, sorts them by score from highest to lowest, and returns the top student’s name and score. Handle majors that are not found.
import pandas as pd

# Original student data
data = {
    'StudentID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve',
             'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Score': [85, 92, 78, 95, 88, 72, 99, 81, 90, 84],
    'Major': ['Computer Science', 'Physics', 'Mathematics',
              'Computer Science', 'Biology', 'Mathematics',
              'Physics', 'Biology', 'Computer Science', 'Mathematics']
}

# Create the DataFrame
student_df = pd.DataFrame(data)

print("Original DataFrame:")
print(student_df)


# Function to find the top student in a major
def find_top_student_in_major(df, major_name):

    # Filter students by the given major
    filtered_df = df[df['Major'] == major_name]

    # Check if the major exists
    if filtered_df.empty:
        print("\nMajor not found:", major_name)
        return None

    # Sort by Score from highest to lowest
    sorted_df = filtered_df.sort_values(
        by='Score',
        ascending=False
    )

    # Select the first student
    top_student = sorted_df.iloc[0]

    # Return only Name and Score
    return top_student[['Name', 'Score']]


# Test 1: Computer Science
print("\nTop Computer Science Student:")
print(find_top_student_in_major(student_df, 'Computer Science'))


# Test 2: Physics
print("\nTop Physics Student:")
print(find_top_student_in_major(student_df, 'Physics'))


# Test 3: Chemistry
print("\nTop Chemistry Student:")
print(find_top_student_in_major(student_df, 'Chemistry'))
```
