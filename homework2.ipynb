```python
#promblem_1
values = [10, 3.14, "Hello", True, 2+3j, None, [1, 2], {"a": 1}]
for value in values:
    result = to_falsy(value)
    print(result)
def to_falsy(value):
    if type(value) == bool:
        return False
    elif type(value) == int:
        return 0
    elif type(value) == float:
        return 0.0
    elif type(value) == str:
        return ""
    elif type(value) == complex:
        return 0j
    elif type(value) == type(None):
        return None
    else:
        print("non primitive type input")
        return None
```

```python
#problem-2
def inspect_operations(arg1, arg2):
    print(type(arg1))
    print(type(arg2))

    try:
        print(arg1 + arg2)
    except TypeError:
        print("cannot add")

    try:
        print(arg1 * arg2)
    except TypeError:
        print("cannot multiply")

    try:
        type(arg2)(arg1)
        print("can coerce")
    except (TypeError, ValueError):
        print("cannot coerce")

challenge_cases = [
    (10.5, 2),
    ("repeat", 4),
    (True, 5),
    ([1, 2], 3),
    ("fizz", "buzz"),
    (None, "a_string"),
    ({'a': 1}, 2)
]

for case in challenge_cases:
    inspect_operations(*case)
    print("-----------")
```
